```python
# Mini test CPU (affiche un progrès pour éviter l'impression "bloqué")
t0 = time.time()
s = 0
n = 20_000_000
step = 5_000_000
for i in range(1, n + 1):
    s += i
    if i % step == 0:
        print('progress', i, '/', n, 'elapsed', round(time.time() - t0, 1), 's')
print('done in', round(time.time() - t0, 2), 's')
```
